```python
# 붓꽃 분류 실습에 필요한 데이터, 모델, 데이터 분할 도구를 불러옴

import sklearn  
from sklearn.datasets import load_iris 
from sklearn.tree import DecisionTreeClassifier 
from sklearn.model_selection import train_test_split 
```

```python
# 붓꽃 데이터를 불러와 입력 특성(iris_data)과 정답 레이블(iris_label)로 나눔

import pandas as pd 

iris = load_iris()  # 붓꽃 데이터와 설명이 담긴 객체를 생성

iris_data = iris.data # 피처 : 꽃받침의 너비, 길이, 꽃잎의 너비, 길이
iris_label = iris.target # 아이리스 품종 : 정답

print("iris target값 :", iris_label)  # 숫자로 표현된 정답들을 출력
print("iris target명 :", iris.target_names)  # 숫자 정답에 대응하는 품종명을 출력
```

```python
iris.data  # 150개 샘플의 4개 특성값을 출력
```

```python
iris.target # 각 붓꽃 샘플의 숫자형 정답 레이블을 확인
```

```python
iris.target_names # 정답 레이블 0, 1, 2에 대응하는 붓꽃 품종 이름을 확인
```

```python
# 특성 배열을 열 이름이 있는 DataFrame으로 만들고 정답 열을 추가

iris_df = pd.DataFrame(data = iris_data, columns= iris.feature_names)  # 특성명으로 열 이름을 지정
iris_df["label"] = iris.target  # 정답 번호를 label 열로 추가
iris_df.head()
```

```python
# 전체 데이터를 학습용 80%와 테스트용 20%로 나눔
# random_state를 고정하면 실행할 때마다 같은 데이터 선택

X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label,
                    test_size = 0.2, random_state = 11)  # 20%를 테스트용으로 사용하고 분할 결과를 고정
```

```python
# 결과를 재현할 수 있도록 난수값을 고정한 결정 트리 모델 생성

dt_clf = DecisionTreeClassifier(random_state=11)
```

```python
# 학습 데이터와 정답을 사용해 결정 트리 모델을 학습

dt_clf.fit(X_train, y_train)
```

```python
# 학습된 모델로 테스트 데이터의 붓꽃 품종을 예측

pred = dt_clf.predict(X_test)
pred
```

```python
# 실제 정답과 예측값을 비교해 정확도를 계산

from sklearn.metrics import accuracy_score  # 분류 정확도 계산 함수를 불러옴
print("예측 정확도 :", accuracy_score(y_test, pred))  # 맞힌 개수를 전체 개수로 나눈 정확도를 출력
```

```python
# 데이터를 5개 구간으로 나누는 K-Fold 객체와 정확도 저장 목록을 준비

from sklearn.model_selection import KFold 
import numpy as np 

kfold = KFold(n_splits=5)  # 데이터를 5개 폴드로 나눔
cv_accuracy = []  # 각 폴드의 정확도를 저장할 빈 리스트
```

```python
# 일반 K-Fold 교차검증을 직접 수행
# 각 회차마다 서로 다른 구간을 검증용으로 사용하고 정확도의 평균을 구함

n_iter = 0  # 현재 교차검증 회차를 셈
features = iris_data  # 입력 특성 배열을 별도 변수로 지정
label = iris.target  # 정답 배열을 별도 변수로 지정

for train_index, test_index in kfold.split(features):  # 폴드마다 학습·검증 행 번호를 받는다.
    X_train, X_test = features[train_index], features[test_index]  # 행 번호로 입력 특성을 나눔
    y_train, y_test = label[train_index], label[test_index]  # 같은 행 번호로 정답을 나눔

    dt_clf.fit(X_train, y_train)  # 현재 폴드의 학습 데이터로 모델을 학습
    pred = dt_clf.predict(X_test)  # 현재 폴드의 검증 데이터를 예측
    n_iter += 1  # 교차검증 회차를 1 증가시킴

    accuracy = np.round(accuracy_score(y_test, pred), 4)  # 정확도를 계산해 소수 넷째 자리로 반올림
    train_size = X_train.shape[0]  # 학습에 사용한 행 수를 구함
    test_size = X_test.shape[0]  # 검증에 사용한 행 수를 구함
    print(n_iter)
    print("교차검증 정확도 :", accuracy)
    print("학습 데이터 크기 :", train_size)
    print("검증 데이터 크기 :", test_size)
    cv_accuracy.append(accuracy)  # 이번 정확도를 평균 계산용 목록에 추가

print("평균 검증 정확도 :", np.mean(cv_accuracy))  # 모든 폴드 정확도의 평균 출력
```

```python
# 품종별 데이터 개수를 확인

iris_df["label"].value_counts()
```

```python
# 일반 K-Fold로 나눴을 때 학습·검증 세트의 품종 분포를 확인

kfold = KFold(n_splits=3)  # 데이터를 순서대로 3개 폴드로 나눔
n_iter = 0  # 현재 분할 회차를 셈

for train_index, test_index in kfold.split(iris_df):  # 각 폴드의 학습·검증 행 번호를 받음
    n_iter += 1  # 분할 회차를 1 증가
    label_train = iris_df["label"].iloc[train_index]  # 학습 행에 해당하는 정답만 선택
    label_test = iris_df["label"].iloc[test_index]  # 검증 행에 해당하는 정답만 선택
    print("학습 레이블 데이터 분포 :", label_train.value_counts())  # 학습 세트의 품종별 개수를 출력
    print("검증 레이블 데이터 분포 :", label_test.value_counts())  # 검증 세트의 품종별 개수를 출력
    print()
```

```python
# StratifiedKFold가 각 세트의 품종 비율을 비슷하게 유지하는지 확인

from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(n_splits=3)  # 품종 비율을 유지하며 3개 폴드로 나눔
n_iter = 0  # 현재 분할 회차를 셈

for train_index, test_index in skf.split(iris_df, iris_df["label"]):  # 정답 비율을 기준으로 행 번호를 나눔
    n_iter += 1  # 분할 회차를 1 증가
    label_train = iris_df["label"].iloc[train_index]  # 학습 세트의 정답을 선택
    label_test = iris_df["label"].iloc[test_index]  # 검증 세트의 정답을 선택
    print("학습 레이블 데이터 분포 :", label_train.value_counts())  # 학습 세트의 품종 비율을 확인
    print("검증 레이블 데이터 분포 :", label_test.value_counts())  # 검증 세트의 품종 비율을 확인
    print()
```

```python
# K-Fold로 결정 트리를 3번 학습·평가하고 평균 정확도를 계산

dt_clf = DecisionTreeClassifier(random_state=156)  # 재현 가능한 결정 트리를 새로 만듬
skfold = StratifiedKFold(n_splits=3)  # 3의 교차검증을 설정
n_iter = 0  # 현재 검증 회차를 셈
cv_accuracy = []  # 회차별 정확도를 저장

for train_index, test_index in skfold.split(features, label):  # 품종 비율을 유지한 행 번호를 받음
    X_train, X_test = features[train_index], features[test_index]  # 입력 특성을 학습·검증용으로 나눔
    y_train, y_test = label[train_index], label[test_index]  # 정답을 학습·검증용으로 나눔
    dt_clf.fit(X_train, y_train)  # 현재 학습 세트로 결정 트리를 학습
    pred = dt_clf.predict(X_test)  # 현재 검증 세트의 품종을 예측

    n_iter += 1  # 검증 회차를 1 증가
    accuracy = np.round(accuracy_score(y_test, pred), 4)  # 현재 폴드 정확도를 계산
    cv_accuracy.append(accuracy)  # 정확도를 목록에 저장

print("평균 검증 정확도 :", np.round(np.mean(cv_accuracy), 4))  # 교차검증의 평균 정확도를 출력
```

```python
# cross_val_score로 위의 교차검증 과정을 간단히 수행

from sklearn.model_selection import cross_val_score

scores = cross_val_score(dt_clf, features, label, scoring = "accuracy", cv = 3)  # 결정 트리를 3번 교차검증
print("교차 검증별 정확도 :", np.round(scores, 4))  # 각 폴드의 정확도를 출력
print("평균 검증 정확도 :", np.round(np.mean(scores), 4))  # 세 정확도의 평균을 출력
```

```python
# 데이터를 나누고 탐색할 결정 트리 옵션을 정의

from sklearn.model_selection import GridSearchCV  

X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target,
                        test_size = 0.2, random_state=121)  # 20%를 테스트용으로 분리하고 결과를 고정

parameters = {"max_depth": [1, 2, 3], "min_samples_split": [2, 3]}  # 탐색할 트리 깊이와 최소 분할 샘플 수
```

```python
# 모든 옵션 조합을 3 교차검증으로 평가하고 결과를 표로 확인

dtree = DecisionTreeClassifier()  # 옵션을 탐색할 기본 결정 트리를 만듭니다.

grid_dtree = GridSearchCV(dtree, param_grid=parameters, cv = 3, refit=True)  # 각 옵션 조합을 3겹 교차검증으로 평가
grid_dtree.fit(X_train, y_train)  # 옵션 탐색과 최적 모델 재학습을 수행
scores_df = pd.DataFrame(grid_dtree.cv_results_)  # 전체 탐색 결과를 DataFrame으로 변환
scores_df[["params", "mean_test_score", "rank_test_score",
           "split0_test_score", "split1_test_score", "split2_test_score"]]  # 옵션·평균·순위·폴드별 점수만 표시
```

```python
grid_dtree.best_params_ # 교차검증에서 가장 높은 성능을 낸 하이퍼파라미터 조합을 확인
```

```python
grid_dtree.best_score_  # 최적 하이퍼파라미터 조합의 평균 교차검증 정확도를 확인
```

```python
# 최적 모델로 테스트 데이터를 예측해 최종 정확도를 확인

estimator = grid_dtree.best_estimator_  # 최적 옵션으로 재학습된 모델 변수 지정

pred = estimator.predict(X_test)  # 최적 모델로 테스트 데이터의 품종을 예측
print(accuracy_score(y_test, pred))  # 최종 테스트 정확도
```

```python
# 붓꽃 특성별 평균과 분산을 확인

from sklearn.datasets import load_iris

iris = load_iris()  # 붓꽃 데이터 객체를 생성
iris_data = iris.data  # 입력 특성 배열 변수 정의
iris_df = pd.DataFrame(data = iris_data, columns=iris.feature_names)  # 특성 배열을 DataFrame으로 변환

print(iris_df.mean())  # 평균을 출력
print()
print(iris_df.var())  # 분산을 출력
```

```python
# StandardScaler로 각 특성의 평균을 0, 분산을 약 1로 표준화
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()  # 표준화 객체를 생성
scaler.fit(iris_df)  # 각 특성의 평균과 표준편차를 계산
iris_scaled = scaler.transform(iris_df)  # 계산된 값으로 모든 특성을 표준화

iris_df_scaled = pd.DataFrame(data = iris_scaled, columns=iris.feature_names)  # 표준화 결과를 DataFrame으로 만듬
print(iris_df_scaled.mean())  # 표준화 후 평균이 0에 가까운지 확인
print()
print(iris_df_scaled.var())  # 표준화 후 분산이 1에 가까운지 확인
```

```python
# MinMaxScaler로 각 특성 값을 0~1 범위로 변환
from sklearn.preprocessing import MinMaxScaler  # 최솟값·최댓값 변환 도구를 불러옴

scaler = MinMaxScaler()  # 0~1 스케일러를 생성
scaler.fit(iris_df)  # 각 특성의 최솟값과 최댓값을 계산
iris_scaled = scaler.transform(iris_df)  # 각 값을 0~1 범위로 변환

iris_df_scaled = pd.DataFrame(data = iris_scaled, columns= iris.feature_names)  # 변환 결과를 DataFrame으로 만듬
print(iris_df_scaled.min())  # 변환 후 열별 최솟값을 확인
print()
print(iris_df_scaled.max())  # 변환 후 열별 최댓값을 확인
```
